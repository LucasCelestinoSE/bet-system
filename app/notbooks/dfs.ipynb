# Implementando algoritmo DFS
1. Crie uma fila contendo todas as pessoas que devem ser verificadas.
2. Retire a pessoa da fila.
3. Confira se a pessoa é um vendedor de mangas
4. Se não, busque todos os vizinhos da pessoa atual e adicione a fila
5. Repita 2 até 4
6. Caso a fila esteja vazia então nao existem vendedores de manga na sua rede.

```python
def pessoa_e_vendedor(nome):
    return nome[-1] == 'm'
```

```python
from collections import deque

def busca_vendedor(nome_inicial):
    fila_de_pesquisa = deque()
    fila_de_pesquisa += grafo[nome_inicial]
    verificados = [] # É importante para não entrar em loop infinito

    while fila_de_pesquisa:
        pessoa = fila_de_pesquisa.popleft()
        
        if not pessoa in verificados: # Só verifica se não olhou antes
            if pessoa_e_vendedor(pessoa):
                print("Encontrado! " + pessoa)
                return True # Finaliza a função COM SUCESSO
            else:
                # Se não for vendedor, adiciona os vizinhos desta pessoa à fila
                fila_de_pesquisa += grafo[pessoa]
                verificados.append(pessoa)
                
    return False 
```

```python
busca_vendedor('lucas')
```
